# Manchester Flood-Aware Pedestrian & Cyclist Routing (MATSim + RoFSW)

Analyse the impact of surface water flooding on non-motorised travel behaviour in Greater Manchester using MATSim routing and Environment Agency RoFSW data.

### Three-Case Comparison:
1. **Case 1 (No rain):** Dry routes evaluated at dry speeds
2. **Case 2 (Uninformed rain):** Same dry routes, but evaluated at flood-affected speeds
3. **Case 3 (Adaptive rain):** Flood-aware routes evaluated at flood speeds

### Data Sources:
- **RoFSW (Risk of Flooding from Surface Water):** UK Environment Agency probabilistic flood dataset
- **MATSim:** Agent-based transport simulation for Greater Manchester
- **OSM:** Road network

```python
# Load, evaluate, and compare 3 flood cases
# Run AFTER ManchesterFloodRouting
import pandas as pd, numpy as np, os

dirs = sorted([d for d in os.listdir(routing_data_path) if d.startswith("results_")])
latest = routing_data_path + dirs[-1] + "/" if dirs else routing_data_path
print("Loading from:", latest)

depths = pd.read_csv(latest + "link_flood_depths.csv")
depth_map = depths.set_index("link_id")["flood_depth"].to_dict()
length_map = depths.set_index("link_id")["length"].to_dict()

def parse_list(s):
    s = s.strip()
    if s.startswith('[') and s.endswith(']'):
        s = s[1:-1]
    if not s.strip():
        return []
    return [x.strip() for x in s.split(',')]

SPEED = {
    "walk": [(0.0, 1.27), (0.1, 1.03), (0.2, 0.85), (0.3, 0.78)],
    "bike": [(0.0, 4.2),  (0.1, 2.04), (0.2, 1.76), (0.3, 1.31)],
}
IMP = {"walk": 0.6, "bike": 0.6}

def flood_speed(mode, depth):
    if depth >= IMP[mode]: return 0.001
    for th, spd in reversed(SPEED[mode]):
        if depth >= th: return spd
    return SPEED[mode][0][1]

def jaccard(a, b):
    inter = len(set(a) & set(b))
    union = len(set(a) | set(b))
    return inter / union if union else 0

def route_flood_metrics(route_links, depth_map, length_map):
    max_d = 0.0
    flooded = 0.0
    for lid in route_links:
        d = depth_map.get(lid, 0.0)
        l = length_map.get(lid, 0.0)
        if d >= 0.1:
            max_d = max(max_d, d)
            flooded += l
    return max_d, flooded

for mode in ["walk", "bike"]:
    dry = pd.read_csv(latest + f"routes_dry_{mode}.csv")
    flood = pd.read_csv(latest + f"routes_flood_{mode}.csv")
    for col in ["links", "path"]:
        dry[col] = dry[col].apply(parse_list)
        flood[col] = flood[col].apply(parse_list)

    rows = []
    for _, row in dry.iterrows():
        t = e = 0.0
        for lid in row["links"]:
            d = depth_map.get(lid, 0.0)
            l = length_map.get(lid, 0.0)
            s = flood_speed(mode, d)
            if s > 0.001:
                t += l / s
                e += d * l / s
            else:
                t += 1e9
                e += 1e9
        rows.append({"person": row["person"], "time_c2": t, "exp_c2": e})
    c2 = pd.DataFrame(rows)

    c1 = dry[["person","travel_time","flood_exposure"]].rename(
        columns={"travel_time": "time_c1", "flood_exposure": "exp_c1"})
    c3 = flood[["person","travel_time","flood_exposure"]].rename(
        columns={"travel_time": "time_c3", "flood_exposure": "exp_c3"})
    merged = c1.merge(c2, on="person").merge(c3, on="person")

    merged["uninformed_delay"] = merged["time_c2"] - merged["time_c1"]
    merged["adaptive_saving"] = merged["time_c2"] - merged["time_c3"]

    merged["jaccard_c2c3"] = merged["person"].apply(
        lambda p: jaccard(
            dry[dry["person"]==p].iloc[0]["path"],
            flood[flood["person"]==p].iloc[0]["path"]))

    # C1: no-flood world, all depths zero
    merged["max_depth_c1"] = 0.0
    merged["flooded_dist_c1"] = 0.0

    # C2: same routes as C1, evaluated on flood network
    c2_metrics = dry["links"].apply(lambda x: route_flood_metrics(x, depth_map, length_map))
    merged["max_depth_c2"] = c2_metrics.apply(lambda x: x[0])
    merged["flooded_dist_c2"] = c2_metrics.apply(lambda x: x[1])

    # C3: flood-aware routes, evaluated on flood network
    c3_metrics = flood["links"].apply(lambda x: route_flood_metrics(x, depth_map, length_map))
    merged["max_depth_c3"] = c3_metrics.apply(lambda x: x[0])
    merged["flooded_dist_c3"] = c3_metrics.apply(lambda x: x[1])

    def to_min(s): return s / 60.0

    impassable = (merged["time_c2"] >= 1e8).sum()

    print()
    print(f"=== {mode}: Paper-ready stats ===")
    print(f"  Trips with impassable links: {impassable} / {len(merged)}")
    print()
    print("--- Table IV: Travel time (min) ---")
    for c, cname in [("time_c1", "Case 1"), ("time_c2", "Case 2"), ("time_c3", "Case 3")]:
        print(f"  {cname}: median={to_min(merged[c].median()):.1f}  mean={to_min(merged[c].mean()):.1f}")
    print(f"  Uninformed delay: median={to_min(merged['uninformed_delay'].median()):.1f}  mean={to_min(merged['uninformed_delay'].mean()):.1f}")
    print(f"  Adaptive saving: median={to_min(merged['adaptive_saving'].median()):.1f}  mean={to_min(merged['adaptive_saving'].mean()):.1f}")
    print()
    print("--- Table V: Flood exposure ---")
    for c, cname in [("c1", "Case 1"), ("c2", "Case 2"), ("c3", "Case 3")]:
        print(f"  {cname}: max_depth mean={merged[f'max_depth_{c}'].mean():.3f}  flooded_dist mean={merged[f'flooded_dist_{c}'].mean():.1f}")
    print()
    print("--- Route divergence ---")
    print(f"  Jaccard C2 vs C3: mean={merged['jaccard_c2c3'].mean():.3f}  median={merged['jaccard_c2c3'].median():.3f}")
```
